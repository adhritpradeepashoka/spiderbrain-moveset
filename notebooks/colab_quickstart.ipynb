```
import numpy as np
from spiderbrain.config import load_config, with_profile
from spiderbrain.generate import generate
from spiderbrain.sim.rate import RateNet
from spiderbrain.sensors.eyes import EyeFrontEnd
cfg=with_profile(load_config(),'default'); c=generate(cfg,0)
net=RateNet(c,backend='auto',sensory_cfg=cfg['regions']['regions']); eye=EyeFrontEnd(c,cfg['eyes'],gain=5)
print(net.backend, net.device)
for t in range(200):
    f=np.zeros((120,160,3),np.float32); f[50:70,10+t//4:20+t//4]=1
    net.step(eye(f))
print('DN left/right', net.r[c.region_ids('dn',0)].mean(), net.r[c.region_ids('dn',1)].mean())
```
